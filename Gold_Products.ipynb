## **DLT Pipeline**

**Streaming Table**

```python
# Expectations
my_rules = {
    "rule1": "product_id IS NOT NULL",
    "rule2": "product_name IS NOT NULL"
}
```

```python
import dlt
from pyspark.sql.functions import *
```

```python
@dlt.table()

@dlt.expect_all_or_drop(my_rules)
def DimProducts_stage():
    df = spark.readStream.table("retail.silver.products_silver")
    return df
```

**Streaming View**

```python
@dlt.view
def DimProducts_view():
    df = spark.readStream.table("Live.DimProducts_stage")
    return df
```

**Dim Products**

```python
dlt.create_streaming_table("DimProducts")
```

```python
dlt.apply_changes(
    target = "DimProducts",
    source = "Live.DimProducts_view",
    keys = ["product_id"],
    sequence_by = "product_id",
    # Optional parameters
    #apply_as_deletes = None,          # Expression to treat an event as a delete (e.g., F.expr("op = 'D'"))
    #apply_as_truncates = None,        # Expression to treat an event as a truncate
    #column_list = None,               # List of columns to include in the target table
    #except_column_list = None,        # List of columns to exclude from updates
    #ignore_null_updates = False,      # Ignore updates with null values
    stored_as_scd_type = "2",         # "1" for SCD Type 1, "2" for SCD Type 2
    #track_history_column_list = None, # Columns to track history for SCD Type 2
    #track_history_except_column_list = None
)
```
